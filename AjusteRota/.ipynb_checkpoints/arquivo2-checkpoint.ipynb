```python
from fpdf import FPDF

class PDF(FPDF):
    def header(self):
        # Arial bold 15
        self.set_font('Arial', 'B', 14)
        # Título
        self.cell(0, 10, 'Relatorio Tecnico: Modelagem Matematica Geoespacial', 0, 1, 'C')
        self.ln(5)

    def footer(self):
        # Posição a 1.5 cm de baixo
        self.set_y(-15)
        # Arial italic 8
        self.set_font('Arial', 'I', 8)
        # Número da página
        self.cell(0, 10, 'Pagina ' + str(self.page_no()), 0, 0, 'C')

    def chapter_title(self, label):
        # Arial 12
        self.set_font('Arial', 'B', 12)
        self.set_fill_color(200, 220, 255)
        self.cell(0, 6, label, 0, 1, 'L', 1)
        self.ln(4)

    def chapter_body(self, body):
        # Times 12
        self.set_font('Times', '', 12)
        self.multi_cell(0, 5, body)
        self.ln()

    def add_formula(self, formula):
        # Fonte Courier para parecer código/matemática
        self.set_font('Courier', '', 11)
        self.set_x(20) # Indentação
        self.multi_cell(0, 5, formula)
        self.ln()

def gerar_pdf():
    pdf = PDF()
    pdf.add_page()
    
    # --- INTRODUÇÃO ---
    pdf.chapter_title('1. Introducao e Desafio')
    texto_intro = (
        "Para explicar matematicamente o calculo da projecao ortogonal de coordenadas geograficas "
        "(latitude/longitude), precisamos dividir o problema em tres fases: Linearizacao (Geodesia), "
        "Algebra Vetorial (Geometria Analitica) e Deslinearizacao.\n\n"
        "O grande desafio e que a Terra e curva (quase uma esfera), mas a geometria vetorial classica "
        "funciona em planos. Como a distancia entre os pontos em Cameta e pequena (menos de 1 km), "
        "podemos usar uma tecnica chamada 'Aproximacao do Plano Tangente Local'."
    )
    pdf.chapter_body(texto_intro)

    # --- FASE 1 ---
    pdf.chapter_title('Fase 1: Transformacao de Espaco (Graus para Metros)')
    texto_fase1 = (
        "Latitude e Longitude sao angulos. Para aplicar vetores, transformamos esses angulos em "
        "distancias lineares (metros) assumindo um raio da Terra R ~ 6.371.000m.\n\n"
        "Definimos os fatores de conversao:"
    )
    pdf.chapter_body(texto_fase1)
    
    formulas_f1 = (
        "k_lat = (2 * pi * R) / 360  [Aprox. 111.320 metros/grau]\n"
        "k_lon = k_lat * cos(latitude_media)"
    )
    pdf.add_formula(formulas_f1)

    pdf.chapter_body(
        "Assumimos o ponto A como a origem (0,0). Para qualquer ponto P(lat, lon), "
        "suas coordenadas cartesianas P_xy(x, y) sao:"
    )
    formulas_f1_b = (
        "y = (lat_P - lat_A) * k_lat\n"
        "x = (lon_P - lon_A) * k_lon"
    )
    pdf.add_formula(formulas_f1_b)

    # --- FASE 2 ---
    pdf.chapter_title('Fase 2: Modelagem Vetorial (Projecao Ortogonal)')
    texto_fase2 = (
        "Agora estamos em um plano 2D (Metros). Temos A=(0,0), C=(xc, yc) e B=(xb, yb).\n"
        "Queremos encontrar P, a projecao de B sobre o vetor AC.\n"
        "Calculamos o 'Fator de Projecao' (t) usando o Produto Escalar:"
    )
    pdf.chapter_body(texto_fase2)

    formulas_f2 = (
        "Vetor u = AC = (xc, yc)\n"
        "Vetor v = AB = (xb, yb)\n\n"
        "Produto Escalar (dot) = (xc * xb) + (yc * yb)\n"
        "Modulo Quadrado Base  = (xc^2 + yc^2)\n\n"
        "t = Produto Escalar / Modulo Quadrado Base"
    )
    pdf.add_formula(formulas_f2)

    pdf.chapter_body(
        "O valor 't' indica a porcentagem do caminho de A ate C. "
        "Aplicamos 't' para achar as coordenadas (xp, yp) do ponto projetado em metros:"
    )
    formulas_f2_b = (
        "xp = t * xc\n"
        "yp = t * yc"
    )
    pdf.add_formula(formulas_f2_b)

    # --- FASE 3 ---
    pdf.chapter_title('Fase 3: Transformacao Inversa (Metros para Graus)')
    texto_fase3 = (
        "Com o ponto P calculado em metros a partir de A, convertemos de volta para "
        "o sistema de coordenadas GPS (Latitude/Longitude) invertendo a logica da Fase 1:"
    )
    pdf.chapter_body(texto_fase3)

    formulas_f3 = (
        "Delta_lat = yp / k_lat\n"
        "Delta_lon = xp / k_lon\n\n"
        "Lat_Final = Lat_A + Delta_lat\n"
        "Lon_Final = Lon_A + Delta_lon"
    )
    pdf.add_formula(formulas_f3)

    # --- RESUMO ---
    pdf.ln(5)
    pdf.set_font('Arial', 'B', 12)
    pdf.cell(0, 6, 'Conclusao', 0, 1, 'L')
    pdf.set_font('Times', '', 12)
    pdf.multi_cell(0, 5, 
        "O script Python executa essas tres fases sequencialmente. "
        "Essa abordagem garante que o angulo de 90 graus seja geometricamente "
        "preciso no mundo real (metros), contornando as distorcoes que ocorreriam "
        "se usassemos graus brutos como unidade de distancia."
    )

    # Salvar
    nome_arquivo = "relatorio_modelagem_cameta.pdf"
    pdf.output(nome_arquivo)
    print(f"Sucesso! O arquivo '{nome_arquivo}' foi gerado na pasta.")

if __name__ == '__main__':
    try:
        gerar_pdf()
    except Exception as e:
        print(f"Erro ao gerar PDF: {e}")
        print("Certifique-se de ter instalado a biblioteca: pip install fpdf")
```
