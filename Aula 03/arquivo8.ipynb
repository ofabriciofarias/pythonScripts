```python
import numpy as np
```

```python
#Criar duas matrizes!
matrizA = np.array([[2,3],[1,4]])
matrizB = np.array([[7,8],[5,6]])
```

```python
#Multiplicação de matrizes
#O número de colunas da primeira tem que ser igual ao número de linhas
#da segunda.
multiplica = np.dot(matrizA, matrizB)
```

```python
multiplica
```

```python
#Outra opção de multiplicação
multiplica2 = matrizA @ matrizB
```
