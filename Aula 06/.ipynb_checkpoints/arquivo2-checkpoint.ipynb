```python
import cv2
import numpy as np
```

```python
caminho_imagem = "bentoespock.jpeg"
```

```python
# Carregamento da imagem original
img = cv2.imread(caminho_imagem)
```

```python
#retira a largura e altura da imagem
altura, largura = img.shape[:2]
```

```python
# ==============================================================================
# 1. Resize (Redimensionamento)
# ==============================================================================
# Redimensiona para metade da resolução original (ou outra dimensão desejada)
img_resize = cv2.resize(
    img, (largura // 2, altura // 2), interpolation=cv2.INTER_LINEAR
)
cv2.imwrite("aug_01_resize.jpeg", img_resize)
```

```python
# ==============================================================================
# 2. Flip (Espelhamento horizontal)
# ==============================================================================
# 1 indica espelhamento horizontal (horizontal flip)
img_flip = cv2.flip(img, 1)
cv2.imwrite("aug_02_flip.jpeg", img_flip)
```

```python
# ==============================================================================
# 3. Rotate (Rotação)
# ==============================================================================
# Rotação de 45 graus em torno do centro da imagem
centro = (largura // 2, altura // 2)
matriz_rotacao = cv2.getRotationMatrix2D(centro, angle=45, scale=1.0)
img_rotate = cv2.warpAffine(img, matriz_rotacao, (largura, altura))
cv2.imwrite("aug_03_rotate.jpeg", img_rotate)
```

```python
# ==============================================================================
# 4. Random Crop (Recorte Aleatório)
# ==============================================================================
# Define o tamanho do recorte (ex.: 70% das dimensões originais)
crop_h = int(altura * 0.7)
crop_w = int(largura * 0.7)
y_ini = np.random.randint(0, altura - crop_h)
x_ini = np.random.randint(0, largura - crop_w)
img_crop = img[y_ini : y_ini + crop_h, x_ini : x_ini + crop_w]
cv2.imwrite("aug_04_random_crop.jpeg", img_crop)
```

```python
# ==============================================================================
# 5. Translation (Translação / Deslocamento)
# ==============================================================================
# Deslocamento em pixels: +60 no eixo X (direita) e +40 no eixo Y (baixo)
matriz_translacao = np.float32([[1, 0, 60], [0, 1, 40]])
img_translation = cv2.warpAffine(img, matriz_translacao, (largura, altura))
cv2.imwrite("aug_05_translation.jpeg", img_translation)
```

```python
# ==============================================================================
# 6. Shearing (Cisalhamento)
# ==============================================================================
# Aplica distorção linear (cisalhamento de 0.2 no eixo X)
matriz_shearing = np.float32([[1, 0.2, 0], [0, 1, 0]])
img_shearing = cv2.warpAffine(img, matriz_shearing, (largura, altura))
cv2.imwrite("aug_06_shearing.jpeg", img_shearing)
```

```python
# ==============================================================================
# 7. Color Jittering (Variação de Brilho, Contraste e Matiz)
# ==============================================================================
# Converte para o espaço HSV para alterar matiz e saturação/brilho
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV).astype(np.float32)
# Altera matiz (H)
hsv[:, :, 0] = (hsv[:, :, 0] + 25) % 180
# Altera saturação (S) e brilho (V)
hsv[:, :, 1] = np.clip(hsv[:, :, 1] * 1.3, 0, 255)
hsv[:, :, 2] = np.clip(hsv[:, :, 2] * 1.2, 0, 255)
img_color_jitter = cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2BGR)
cv2.imwrite("aug_07_color_jittering.jpeg", img_color_jitter)
```

```python
# ==============================================================================
# 8. Random Erasing (Apagamento Aleatório de Região)
# ==============================================================================
# Cria uma máscara/bloco retangular cinza aleatório simulando oclusão
img_random_erasing = img.copy()
area_h = int(altura * 0.25)
area_w = int(largura * 0.25)
y_erase = np.random.randint(0, altura - area_h)
x_erase = np.random.randint(0, largura - area_w)
img_random_erasing[
    y_erase : y_erase + area_h, x_erase : x_erase + area_w
] = np.random.randint(0, 256, (area_h, area_w, 3), dtype=np.uint8)
cv2.imwrite("aug_08_random_erasing.jpeg", img_random_erasing)
```

```python
# ==============================================================================
# 9. Gaussian Noise (Ruído Gaussiano)
# ==============================================================================
# Adiciona ruído com distribuição normal
gaussiano = np.random.normal(0, 25, img.shape)
img_gaussian_noise = np.clip(img.astype(np.float32) + gaussiano, 0, 255).astype(
    np.uint8
)
cv2.imwrite("aug_09_gaussian_noise.jpeg", img_gaussian_noise)
```

```python
# ==============================================================================
# 10. Mixup (Mistura com outra imagem)
# ==============================================================================
# Cria a segunda imagem (aqui usamos a imagem espelhada como exemplo da 2ª classe/amostra)
img2 = cv2.flip(img, 0)
lambda_mix = 0.6  # Peso da imagem original (60% da original + 40% da segunda)
img_mixup = cv2.addWeighted(img, lambda_mix, img2, 1 - lambda_mix, 0)
cv2.imwrite("aug_10_mixup.jpeg", img_mixup)
```
