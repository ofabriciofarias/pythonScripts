```python
import os
import cv2

# ==============================================================================
# Configurações Iniciais
# ==============================================================================
caminho_imagem = "bentoespock.jpeg"
id_classe = 1  # 1 representa a classe 'gato'

# Carregamento da imagem original
img_original = cv2.imread(caminho_imagem)

# Cópia para exibição contínua na interface
img_canvas = img_original.copy()
altura_img, largura_img = img_original.shape[:2]

# Variáveis globais para rastreamento do mouse
desenhando = False
ix, iy = -1, -1
bounding_boxes = []  # Lista para armazenar as caixas em pixels [(x1, y1, x2, y2), ...]


# ==============================================================================
# Função de Callback do Mouse
# ==============================================================================
def marcar_bounding_box(event, x, y, flags, param):
    global ix, iy, desenhando, img_canvas, bounding_boxes

    # 1. Pressionou o botão esquerdo: Início da Bounding Box
    if event == cv2.EVENT_LBUTTONDOWN:
        desenhando = True
        ix, iy = x, y

    # 2. Movimento do mouse com botão pressionado: Pré-visualização dinâmica
    elif event == cv2.EVENT_MOUSEMOVE:
        if desenhando:
            # Restaura o canvas com as caixas fixadas para não acumular rastros
            img_canvas = img_original.copy()
            for b_x1, b_y1, b_x2, b_y2 in bounding_boxes:
                cv2.rectangle(
                    img_canvas, (b_x1, b_y1), (b_x2, b_y2), (0, 255, 0), 2
                )
                cv2.putText(
                    img_canvas,
                    "gato",
                    (b_x1, max(b_y1 - 5, 15)),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.5,
                    (0, 255, 0),
                    1,
                )

            # Desenha a caixa temporária em amarelo durante o arraste
            cv2.rectangle(img_canvas, (ix, iy), (x, y), (0, 255, 255), 1)

    # 3. Soltou o botão esquerdo: Conclusão e registro da Bounding Box
    elif event == cv2.EVENT_LBUTTONUP:
        desenhando = False

        # Garante que x1 < x2 e y1 < y2 independente da direção em que foi arrastado
        x1, x2 = min(ix, x), max(ix, x)
        y1, y2 = min(iy, y), max(iy, y)

        # Evita cliques acidentais sem deslocamento mínimo (área nula)
        if (x2 - x1) > 3 and (y2 - y1) > 3:
            bounding_boxes.append((x1, y1, x2, y2))

        # Atualiza a imagem com a caixa final em verde
        img_canvas = img_original.copy()
        for b_x1, b_y1, b_x2, b_y2 in bounding_boxes:
            cv2.rectangle(
                img_canvas, (b_x1, b_y1), (b_x2, b_y2), (0, 255, 0), 2
            )
            cv2.putText(
                img_canvas,
                "gato",
                (b_x1, max(b_y1 - 5, 15)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (0, 255, 0),
                1,
            )


# ==============================================================================
# Loop Principal da Interface (HighGUI)
# ==============================================================================
nome_janela = (
    "Anotador YOLO - Arraste para marcar os gatos | ESC: Salvar e Sair"
)
cv2.namedWindow(nome_janela)
cv2.setMouseCallback(nome_janela, marcar_bounding_box)

print("--- INSTRUÇÕES ---")
print("1. Clique e arraste com o botão esquerdo do mouse sobre cada gato.")
print("2. Solte para confirmar a caixa.")
print("3. Pressione a tecla 'ESC' para concluir e exportar o arquivo YOLO.")

while True:
    cv2.imshow(nome_janela, img_canvas)
    tecla = cv2.waitKey(20) & 0xFF

    # Tecla ESC (ASCII 27) para sair e gerar o arquivo
    if tecla == 27:
        break

cv2.destroyAllWindows()

# ==============================================================================
# Conversão e Exportação para o Formato Padrão YOLO (.txt)
# ==============================================================================
nome_base = os.path.splitext(os.path.basename(caminho_imagem))[0]
arquivo_saida = f"{nome_base}.txt"

linhas_yolo = []
for x1, y1, x2, y2 in bounding_boxes:
    # 1. Dimensões em pixels
    box_w = x2 - x1
    box_h = y2 - y1
    x_centro = x1 + (box_w / 2.0)
    y_centro = y1 + (box_h / 2.0)

    # 2. Normalização relativa às dimensões totais da imagem ([0.0, 1.0])
    x_center_norm = x_centro / largura_img
    y_center_norm = y_centro / altura_img
    w_norm = box_w / largura_img
    h_norm = box_h / altura_img

    # Formatação YOLO: <classe> <x_center> <y_center> <width> <height>
    linhas_yolo.append(
        f"{id_classe} {x_center_norm:.6f} {y_center_norm:.6f} {w_norm:.6f} {h_norm:.6f}\n"
    )

with open(arquivo_saida, "w", encoding="utf-8") as f:
    f.writelines(linhas_yolo)

print(f"\nAnotação concluída!")
print(f"Total de caixas marcadas: {len(bounding_boxes)}")
print(f"Arquivo gerado: '{arquivo_saida}'")
```
