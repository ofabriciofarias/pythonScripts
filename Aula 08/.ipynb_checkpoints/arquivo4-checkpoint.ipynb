```python
import cv2
import numpy as np

# -------------------------------------------------------------------------
# Carregamento da imagem base
# -------------------------------------------------------------------------
caminho_imagem = "bentoespock.jpeg"

# Leitura direta em tons de cinza para operações morfológicas de intensidade
img_gray = cv2.imread(caminho_imagem, cv2.IMREAD_GRAYSCALE)

if img_gray is None:
    raise FileNotFoundError(
        f"Não foi possível encontrar a imagem '{caminho_imagem}'. Verifique o caminho/nome."
    )

# Definição do Elemento Estruturante (Kernel 5x5 retangular)
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# =========================================================================
# 1. Erosão
# =========================================================================
# Substitui o pixel central pelo valor mínimo sob o kernel
img_erosao = cv2.erode(img_gray, kernel, iterations=1)
cv2.imwrite("morfologia_01_erosao.jpeg", img_erosao)

# =========================================================================
# 2. Dilatação
# =========================================================================
# Substitui o pixel central pelo valor máximo sob o kernel
img_dilatacao = cv2.dilate(img_gray, kernel, iterations=1)
cv2.imwrite("morfologia_02_dilatacao.jpeg", img_dilatacao)

# =========================================================================
# 3. Abertura (Erosão seguida de Dilatação)
# =========================================================================
# Remove ruídos claros menores que o kernel e suaviza o contorno
img_abertura = cv2.morphologyEx(img_gray, cv2.MORPH_OPEN, kernel)
cv2.imwrite("morfologia_03_abertura.jpeg", img_abertura)

# =========================================================================
# 4. Fechamento (Dilatação seguida de Erosão)
# =========================================================================
# Fecha pequenas falhas e orifícios escuros
img_fechamento = cv2.morphologyEx(img_gray, cv2.MORPH_CLOSE, kernel)
cv2.imwrite("morfologia_04_fechamento.jpeg", img_fechamento)

# =========================================================================
# 5. Filtragem Morfológica
# =========================================================================
# 5.1 Gradiente Morfológico: Destaca bordas (Dilatação - Erosão)
img_gradiente = cv2.morphologyEx(img_gray, cv2.MORPH_GRADIENT, kernel)
cv2.imwrite("morfologia_05_gradiente.jpeg", img_gradiente)

# 5.2 Top-Hat: Realça elementos mais claros que a vizinhança (Original - Abertura)
img_tophat = cv2.morphologyEx(img_gray, cv2.MORPH_TOPHAT, kernel)
cv2.imwrite("morfologia_06_tophat.jpeg", img_tophat)

# 5.3 Black-Hat: Realça elementos mais escuros que a vizinhança (Fechamento - Original)
img_blackhat = cv2.morphologyEx(img_gray, cv2.MORPH_BLACKHAT, kernel)
cv2.imwrite("morfologia_07_blackhat.jpeg", img_blackhat)

print("Todas as operações morfológicas foram processadas e salvas com sucesso.")
```
