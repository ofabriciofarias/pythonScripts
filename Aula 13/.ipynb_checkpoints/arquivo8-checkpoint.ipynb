```python
import cv2
import numpy as np

def detectar_e_desenhar_cantos_harris(frame_bgr, limiar_harris_percentual=0.01):
    """
    Aplica o Harris Corner Detector em um frame e desenha os cantos detectados.

    Parâmetros:
    - frame_bgr (numpy.ndarray): O frame de vídeo no formato BGR.
    - limiar_harris_percentual (float): A proporção do valor máximo da matriz Harris
      que define o limiar para um ponto ser considerado um "canto". (e.g., 0.01 = 1%)

    Retorna:
    - int: O número de cantos detectados que passaram do limiar.
    - numpy.ndarray: O frame com os cantos marcados.
    """
    
    # 1. Converter para escala de cinza
    imagem_cinza = cv2.cvtColor(frame_bgr, cv2.COLOR_BGR2GRAY)
    
    # 2. Converter para float32 (necessário para cv2.cornerHarris)
    imagem_cinza_float = np.float32(imagem_cinza) 

    # 3. Aplicar Harris Corner Detector
    # cv2.cornerHarris(src, blockSize, ksize, k)
    # blockSize: Vizinhança para consideração do detector (geralmente 2)
    # ksize: Tamanho do kernel Sobel (geralmente 3)
    # k: Parâmetro Harris (geralmente 0.04)
    matriz_harris = cv2.cornerHarris(imagem_cinza_float, 2, 3, 0.04)
    
    # 4. Dilatar a matriz de resposta (Opcional: Aumenta a visibilidade dos cantos)
    matriz_harris_dilatada = cv2.dilate(matriz_harris, None)
    
    # 5. Definir o limiar para detectar cantos fortes
    # O limiar é uma porcentagem (limiar_harris_percentual) do valor máximo da matriz Harris.
    limiar = limiar_harris_percentual * matriz_harris_dilatada.max()
    
    # 6. Identificar os pixels que são "cantos fortes"
    # numpy.where retorna as coordenadas (y, x) onde a condição é verdadeira.
    coordenadas_cantos = np.where(matriz_harris_dilatada > limiar)
    
    # 7. Desenhar os cantos no frame original
    imagem_resultado = frame_bgr.copy()
    
    # As coordenadas_cantos são (array_y, array_x). Usamos zip para emparelhá-las.
    for y, x in zip(*coordenadas_cantos):
        # Desenha um círculo no local do canto.
        cv2.circle(imagem_resultado, (x, y), 3, (0, 0, 255), -1) # Círculo vermelho sólido
        
    # O número total de cantos detectados é o tamanho do array de coordenadas Y (ou X)
    num_cantos = len(coordenadas_cantos[0])

    return num_cantos, imagem_resultado

```

```python
## 💻 Main Loop Logic

# --- Lógica Principal (Captura e Contagem) ---
# Captura do video
cap = cv2.VideoCapture(0)

# Criar uma janela para o video
cv2.namedWindow('Harris Corner Detector')

# Variável de estado para a contagem total/acumulada
contador_total = 0 
# Contador do frame anterior, usado para verificar se novos cantos surgiram.
contagem_anterior = 0 

while True:
    
    # Capturar frame a frame
    ret, frame = cap.read()
    if not ret:
        break

    # 1. Detectar e contar os cantos no frame ATUAL
    # O parâmetro '0.01' (1%) é o limiar de sensibilidade do detector Harris.
    contagem_atual, resultado_frame = detectar_e_desenhar_cantos_harris(
        frame, 
        limiar_harris_percentual=0.01 
    )

    # 2. Aplicar a lógica de contagem acumulada (adaptada)
    # Nota: A contagem de cantos (features) é inerentemente mais ruidosa que a contagem de objetos (contornos).
    # Se você tiver 4 cantos por objeto, a contagem aumentará em 4x.
    
    # Para simplificar, vamos contar apenas o número de "cantos fortes" detectados no frame.
    # Se a contagem aumentar significativamente, assumimos que um novo objeto com cantos entrou na vista.
    
    # A lógica original do seu código:
    if contagem_atual > contagem_anterior:
        # Se o número de features AUMENTOU, adicionamos a diferença.
        cantos_novos = contagem_atual - contagem_anterior
        contador_total += cantos_novos
    
    # Se a contagem diminuir muito drasticamente, podemos considerar que o objeto saiu, 
    # mas a remoção de features é mais instável que a remoção de contornos.
    
    if contador_total < 0:
        contador_total = 0

    # 3. Atualizar a contagem anterior para o próximo loop
    contagem_anterior = contagem_atual

    # 4. Mostrar o resultado
    cv2.putText(resultado_frame, f"TOTAL ACUMULADO de Cantos (Features): {contador_total}", 
                (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2) # Amarelo
    cv2.putText(resultado_frame, f"Cantos no Frame: {contagem_atual}", 
                (10, 70), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 255), 2) # Ciano

    # Apresentar o frame resultante
    cv2.imshow('Harris Corner Detector', resultado_frame)
    
    # Comando de saída
    if cv2.waitKey(1) & 0xFF == ord('s'):
        break
    
# Quando finalizar, destruir os elementos
cap.release()
cv2.destroyAllWindows()
```
