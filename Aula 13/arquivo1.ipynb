```python
#Harris Corner Detection
#Para identificação de bordas na imagem
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
xadrez = cv2.imread('xadrez.jpg')
xadrez = cv2.cvtColor(xadrez, cv2.COLOR_BGR2RGB)
```

```python
xadrez
```

```python
plt.imshow(xadrez)
plt.show()
```

```python
cinza_xadrez = cv2.cvtColor(xadrez, cv2.COLOR_RGB2GRAY)
```

```python
cinza_xadrez
```

```python
plt.imshow(cinza_xadrez, cmap = 'gray')
plt.show()
```

```python
xadrez_real = cv2.imread('xadrez_real.jpg')
xadrez_real = cv2.cvtColor(xadrez_real, cv2.COLOR_BGR2RGB)
plt.imshow(xadrez_real)
plt.show()
```

```python
cinza_xadrez_real = cv2.cvtColor(xadrez_real, cv2.COLOR_RGB2GRAY)
plt.imshow(cinza_xadrez_real, cmap = 'gray')
plt.show()
```

```python
cinza_xadrez
```

```python
cinza = np.float32(cinza_xadrez)
```

```python
cinza
```

```python
#Aplicando o algoritmo Harris Corner Detection
destino = cv2.cornerHarris(src = cinza, blockSize = 2, ksize = 3, k = 0.04)
```

```python
#Transformação morfológica pelo cv2.dilate()
destino = cv2.dilate(destino, None)
```

```python
xadrez[destino > 0.02*destino.max()] = [0, 255, 0]
```

```python
imagem_plana = cv2.imread('imagem_plana.jpg')
imagem_plana = cv2.cvtColor(imagem_plana, cv2.COLOR_BGR2RGB)
plt.imshow(imagem_plana)
plt.show()

imagem_plana_cinza = cv2.cvtColor(imagem_plana, cv2.COLOR_RGB2GRAY)
plt.imshow(imagem_plana_cinza, cmap = 'gray')
plt.show()
```

```python
cinza = np.float32(imagem_plana_cinza)
destino = cv2.cornerHarris(src = cinza, blockSize = 2, ksize = 3, k = 0.04)
destino2 = cv2.dilate(destino, None)
imagem_plana[destino2 > 0.2 * destino2.max()] = [255,0,0]
plt.imshow(imagem_plana)
plt.show()
```

```python
imagem = cv2.imread('imagem_bordas.jpg')
imagem = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)
plt.imshow(imagem)
plt.show()

imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_RGB2GRAY)
plt.imshow(imagem_cinza, cmap = 'gray')
plt.show()

cinza = np.float32(imagem_cinza)
destino = cv2.cornerHarris(src = cinza, blockSize = 2, ksize = 3, k = 0.04)
destino = cv2.dilate(destino, None)
imagem[destino > 0.0001 * destino.max()] = [255,0,0]
plt.imshow(imagem)
plt.show()
```

```python
imagem = cv2.imread('bola.png')
imagem = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)
plt.imshow(imagem)
plt.show()

imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_RGB2GRAY)
plt.imshow(imagem_cinza, cmap = 'gray')
plt.show()

cinza = np.float32(imagem_cinza)
destino = cv2.cornerHarris(src = cinza, blockSize = 2, ksize = 3, k = 0.04)
destino = cv2.dilate(destino, None)
imagem[destino > 0.001 * destino.max()] = [255,0,0]
plt.imshow(imagem)
plt.show()
```
