```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load and convert the image to grayscale
img = cv2.imread('xadrez.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Convert to float32 as required by the function
gray = np.float32(gray)

# 2. Apply the Harris corner detector
# blockSize: Neighborhood size (e.g., 2)
# ksize: Sobel kernel size (e.g., 3)
# k: Harris detector free parameter (e.g., 0.04)
dst = cv2.cornerHarris(gray, 2, 3, 0.04)

# 3. Dilate the corner points for better visibility
dst = cv2.dilate(dst, None)

# 4. Threshold the result and mark corners on the original image
# Corners are marked in red where the response is above 1% of the maximum
img[dst > 0.01 * dst.max()] = [0, 0, 255] # BGR color for red

# Display the result
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Harris Corner Detection')
plt.axis('off')
plt.show()
```
