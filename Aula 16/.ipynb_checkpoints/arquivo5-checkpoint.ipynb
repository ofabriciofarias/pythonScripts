```python
import cv2

# 1. Initialize the HOG descriptor
hog = cv2.HOGDescriptor()

# 2. Set the Support Vector Machine (SVM) to be the default people detector
hog.setSVMDetector(cv2.HOGDescriptor_getDefaultPeopleDetector())

# Load image
image = cv2.imread('pedestres.jpg')
# Native OpenCV equivalent
height, width = image.shape[:2]
target_width = 800
ratio = target_width / float(width)
new_height = int(height * ratio)
image = cv2.resize(image, (target_width, new_height))

# 3. Detect people
# returns bounding boxes (rects) and weights
(rects, weights) = hog.detectMultiScale(
    image, 
    winStride=(4, 4), 
    padding=(8, 8), 
    scale=1.05
)

# Draw boxes
for (x, y, w, h) in rects:
    cv2.rectangle(image, (x, y), (x + w, y + h), (0, 0, 255), 2)

cv2.imshow("Pedestrians", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
