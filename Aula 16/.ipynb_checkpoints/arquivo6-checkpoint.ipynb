```python
import os
import cv2
import numpy as np

# -------------------------------------------------------------------------
# 1. Carregamento da Imagem
# -------------------------------------------------------------------------
caminho_imagem = "turma.jpg"
if not os.path.exists(caminho_imagem):
    caminho_imagem = "bentoespock_3.jpeg"

img_bgr = cv2.imread(caminho_imagem)
if img_bgr is None:
    raise FileNotFoundError(f"Imagem '{caminho_imagem}' não encontrada.")

# Redimensiona para resolução didática uniforme para a demonstração
img_base = cv2.resize(img_bgr, (640, 480))
img_cinza = cv2.cvtColor(img_base, cv2.COLOR_BGR2GRAY)

# =========================================================================
# PARTE A: Demonstração Conceitual da Janela Deslizante (Sliding Window)
# =========================================================================
canvas_janelas = img_base.copy()
tamanho_janela = (80, 80)
passo = 30  # Stride de 30 pixels

contador_janelas = 0
altura, largura = img_base.shape[:2]

# Simulação da varredura sistemática
for y in range(0, altura - tamanho_janela[1], passo):
    for x in range(0, largura - tamanho_janela[0], passo):
        contador_janelas += 1
        # Desenha apenas algumas janelas em cinza claro para ilustrar a grade de varredura
        if contador_janelas % 3 == 0:
            cv2.rectangle(
                canvas_janelas, 
                (x, y), 
                (x + tamanho_janela[0], y + tamanho_janela[1]), 
                (180, 180, 180), 
                1
            )

texto_aviso = f"Sliding Window: {contador_janelas} recortes gerados (em 1 escala)"
cv2.putText(canvas_janelas, texto_aviso, (15, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)
print(f"[Parte A] Total de janelas avaliadas na varredura simples: {contador_janelas}")

# =========================================================================
# PARTE B: Detecção Rápida via Cascata (Haar Cascade Classifier)
# =========================================================================
# O OpenCV traz classificadores Haar pré-treinados em formato XML
caminho_cascade = cv2.CascadeClassifier('data/haarcascades/haarcascade_frontalface_default.xml') #cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
classificador_face = cv2.CascadeClassifier(caminho_cascade)

# Se o XML frontal falhar para animais, usamos a cascata de perfil/olhos ou o detector dedicado
if classificador_face.empty():
    print("Aviso: Classificador Haar padrão não carregado.")

# detectMultiScale aplica a cascata com pirâmide de escala interna
# scaleFactor=1.1: escala a imagem em 10% a cada nível da pirâmide
# minNeighbors=4: critério de agrupamento (NMS simplificado) para filtrar falsos positivos
faces_detectadas = classificador_face.detectMultiScale(
    img_cinza,
    scaleFactor=1.1,
    minNeighbors=4,
    minSize=(60, 60)
)

canvas_haar = img_base.copy()
for (x, y, w, h) in faces_detectadas:
    # Bounding Box final resultante da cascata
    cv2.rectangle(canvas_haar, (x, y), (x + w, y + h), (0, 255, 0), 2)
    cv2.putText(
        canvas_haar, 
        "Deteccao Cascade", 
        (x, y - 10), 
        cv2.FONT_HERSHEY_SIMPLEX, 
        0.5, 
        (0, 255, 0), 
        1
    )

print(f"[Parte B] Regiões que sobreviveram aos estágios da cascata: {len(faces_detectadas)}")

# =========================================================================
# Exibição dos Resultados Didáticos
# =========================================================================
cv2.imshow("1. Grade de Varredura da Janela Deslizante", canvas_janelas)
cv2.imshow("2. Resultado Acelerado por Cascata (Haar / NMS)", canvas_haar)

print("\nPressione qualquer tecla para encerrar...")
cv2.waitKey(0)
cv2.destroyAllWindows()
```
