```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
objeto = cv2.imread('barcelona.jpg')
objeto_cinza = cv2.cvtColor(objeto, cv2.COLOR_BGR2GRAY)

objeto = cv2.cvtColor(objeto, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(objeto_cinza, cmap= 'gray')
plt.show()
```

```python
objeto_cascade = cv2.CascadeClassifier('data/haarcascades/haarcascade_eye.xml')
```

```python
def objeto_detector(imagem):
    cont = 0
    copia_imagem = imagem.copy()
    
    objeto_retangulo = objeto_cascade.detectMultiScale(copia_imagem, scaleFactor = 1.2, minNeighbors = 5)
    
    for (x,y,w,h) in objeto_retangulo:
        cv2.rectangle(copia_imagem, (x,y), (x+w, y+h), (255,0,0), 10)
        cont = cont + 1
        
    print(cont)
    return copia_imagem
```

```python
resultado = objeto_detector(objeto)
plt.imshow(resultado)
plt.show()
```
