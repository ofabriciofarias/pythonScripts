```python
import torch
import psutil
from ultralytics import YOLO

try:
    from pynvml import (
        nvmlInit,
        nvmlDeviceGetHandleByIndex,
        nvmlDeviceGetMemoryInfo,
        nvmlDeviceGetName,
        nvmlShutdown,
    )
    pynvml_available = True
except ImportError:
    pynvml_available = False

# =========================================
# Treinamento YOLOv8 com detecção automática de GPU
# =========================================

# Detectar se há GPU disponível
if torch.cuda.is_available():
    device = "cuda"
    gpu_name = torch.cuda.get_device_name(0)
    print(f"✅ GPU detectada: {gpu_name}")
else:
    device = "cpu"
    print("⚠️ Nenhuma GPU detectada. O treinamento será feito na CPU.")

# Carregar modelo YOLOv8n
model = YOLO("yolov8n.pt")

# Mostrar memória GPU antes do treino (se disponível)
if pynvml_available and device == "cuda":
    nvmlInit()
    handle = nvmlDeviceGetHandleByIndex(0)
    info = nvmlDeviceGetMemoryInfo(handle)
    print(f"🔹 Memória GPU antes do treino: {info.used / 1024**2:.2f} MB usada de {info.total / 1024**2:.2f} MB totais")
    nvmlShutdown()

# Treinamento
#http://localhost:8888/lab/tree/Documents/python/Yolo/fallyolov8
results = model.train(
    
    data="C:/Users/fabri/Documents/Visao Computacional/Libras.v5i.yolov8/data.yaml",     # Caminho do dataset YAML
    epochs=50,                # Épocas
    imgsz=640,                 # Tamanho da imagem
    batch=16,                  # Batch size
    device=device,             # GPU ou CPU
    name="yolo_simples_treino" # Nome do experimento
)

# Mostrar uso de GPU após o treino (se disponível)
if pynvml_available and device == "cuda":
    nvmlInit()
    handle = nvmlDeviceGetHandleByIndex(0)
    info = nvmlDeviceGetMemoryInfo(handle)
    print(f"🔹 Memória GPU após treino: {info.used / 1024**2:.2f} MB usada de {info.total / 1024**2:.2f} MB totais")
    nvmlShutdown()

print("\n✅ Treinamento concluído!")
print("📂 Resultados salvos em: runs/detect/yolo_simples_treino")
```
