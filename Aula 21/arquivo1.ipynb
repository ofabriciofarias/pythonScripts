```python
from ultralytics import YOLO

# 1. Load a model
# 'yolo11n.pt' is the Nano model (fastest, lightest).
# It will automatically download the weights file on the first run.
model = YOLO('yolo11n.pt') 

# 2. Run inference on a source
# You can use a local image path, a URL, or '0' for webcam.
source = 'https://ultralytics.com/images/bus.jpg' 

# 3. Predict
results = model(source)

# 4. Process results
for result in results:
    boxes = result.boxes  # Boxes object for bbox outputs
    result.show()         # Display result window
    result.save(filename='result.jpg') # Save to disk
```
