```python
import cv2
from ultralytics import YOLO

# 1. Load a model
# 'yolo11n.pt' is the Nano model (fastest, lightest).
# It will automatically download the weights file on the first run.
model = YOLO('yolo11n.pt') 

# 2. Run inference on a source
# You can use a local image path, a URL, or '0' for webcam.

# Captura do video
cap = cv2.VideoCapture(0)

# Check if camera opened successfully
if not cap.isOpened():
    print("Erro: Não foi possível abrir a webcam.")
    exit()
    
while True:
    
    # Capturar frame a frame
    ret, frame = cap.read()
    if not ret:
        break
   
    # 3. Predict
    # We remove 'show=True' so we can handle the window ourselves.
    # We use stream=True to reduce memory usage in loops (optional but recommended).
    results = model(frame)

    # 4. Draw the boxes
    # results is a list. We take the first result (current frame)
    # and use .plot() to get the image with boxes drawn on it.
    annotated_frame = results[0].plot()

    print(results[0])

    # 5. Display the result using OpenCV
    cv2.imshow("Camera YOLO", annotated_frame)
    # Comando de saída
    if cv2.waitKey(1) & 0xFF == ord('s'):
        break
    
# Quando finalizar, destruir os elementos
cap.release()
cv2.destroyAllWindows()
```
