```python
#Sistema de Libras
#Treinado!
import cv2
import torch
from ultralytics import YOLO
import time

# =========================================
# DETECÇÃO EM TEMPO REAL COM A WEBCAM (YOLOv8)
# =========================================

# 1. Detectar se há GPU disponível
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"✅ Dispositivo em uso: {device.upper()}")

# 2. Carregar o modelo treinado
# Substitua o caminho se necessário
model = YOLO("runs/detect/yolo_simples_treino2/weights/best.pt")
model.to(device)

# 3. Iniciar a captura da webcam
cap = cv2.VideoCapture(0)  # 0 = webcam padrão
if not cap.isOpened():
    raise IOError("❌ Não foi possível acessar a câmera.")

print("🎥 Pressione 'q' para sair.\n")

# 4. Loop de captura e inferência
while True:
    ret, frame = cap.read()
    if not ret:
        print("❌ Falha ao capturar frame da câmera.")
        break

    # Reduzir tamanho para melhorar desempenho
    resized_frame = cv2.resize(frame, (640, 480))

    # Fazer a predição com YOLOv8
    results = model.predict(resized_frame, imgsz=640, device=device, verbose=False)

    # 5. Mostrar as detecções no frame
    annotated_frame = results[0].plot()

    # 6. Extrair as classes detectadas
    for box in results[0].boxes:
        cls_id = int(box.cls[0])
        conf = float(box.conf[0])
        label = model.names[cls_id]
        print(f"🧩 Classe: {label} | Confiança: {conf:.2f}")

    # 7. Mostrar o vídeo com anotações
    cv2.imshow("YOLOv8 - Libras", annotated_frame)

    # Pressionar 'q' para sair
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

# 8. Encerrar
cap.release()
cv2.destroyAllWindows()
print("🛑 Execução encerrada.")
```
