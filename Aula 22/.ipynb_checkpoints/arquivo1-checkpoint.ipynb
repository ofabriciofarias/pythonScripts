```python
import os
import cv2
import numpy as np

# 1. Carregamento da Imagem
caminho_imagem = "bentoespock.jpeg"
if not os.path.exists(caminho_imagem):
    caminho_imagem = "bentoespock_3.jpeg"

img_bgr = cv2.imread(caminho_imagem)
if img_bgr is None:
    raise FileNotFoundError(
        f"Não foi possível carregar a imagem. Verifique o caminho/nome do arquivo."
    )

# Conversão para escala de cinza para aplicação dos filtros passa-alta/passa-baixa
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

# ==============================================================================
# FILTROS PASSA-BAIXA (SUAVIZAÇÃO)
# ==============================================================================

# 1. Filtro de Média (Box Filter) - Kernel 9x9
# Cada pixel recebe a média aritmética dos 81 vizinhos
img_media = cv2.blur(img_gray, (9, 9))
cv2.imwrite("filtro_01_media.jpeg", img_media)

# 2. Filtro Gaussiano - Kernel 9x9 com desvio padrão (sigma) calculado automaticamente
# Pesos decrescem exponencialmente do centro para as bordas conforme a curva de Gauss
img_gauss = cv2.GaussianBlur(img_gray, (9, 9), sigmaX=0)
cv2.imwrite("filtro_02_gaussiano.jpeg", img_gauss)

# ==============================================================================
# FILTROS PASSA-ALTA (REALCE E DETECÇÃO DE BORDAS)
# ==============================================================================
# Obs.: As derivadas produzem gradientes negativos e positivos.
# Usamos cv2.CV_64F para não truncar valores negativos em zero,
# tiramos o valor absoluto e convertemos de volta para uint8 [0, 255].

# 3. Filtro de Sobel (Primeira Derivada)
# 3.1 Sobel Horizontal (detecta variações no eixo X / arestas verticais)
sobel_x = cv2.Sobel(img_gauss, cv2.CV_64F, 1, 0, ksize=3)
sobel_x_abs = cv2.convertScaleAbs(sobel_x)
cv2.imwrite("filtro_03_sobel_x.jpeg", sobel_x_abs)

# 3.2 Sobel Vertical (detecta variações no eixo Y / arestas horizontais)
sobel_y = cv2.Sobel(img_gauss, cv2.CV_64F, 0, 1, ksize=3)
sobel_y_abs = cv2.convertScaleAbs(sobel_y)
cv2.imwrite("filtro_04_sobel_y.jpeg", sobel_y_abs)

# 3.3 Magnitude Total do Gradiente de Sobel: sqrt(Gx^2 + Gy^2)
sobel_total = cv2.magnitude(sobel_x, sobel_y)
sobel_total_abs = cv2.convertScaleAbs(sobel_total)
cv2.imwrite("filtro_05_sobel_total.jpeg", sobel_total_abs)

# 4. Filtro Laplaciano (Segunda Derivada)
# Realça transições rápidas isotrópicas (todas as direções simultaneamente)
laplaciano = cv2.Laplacian(img_gauss, cv2.CV_64F, ksize=3)
laplaciano_abs = cv2.convertScaleAbs(laplaciano)
cv2.imwrite("filtro_06_laplaciano.jpeg", laplaciano_abs)

# ==============================================================================
# Exibição dos Resultados em Janelas
# ==============================================================================
cv2.imshow("Original em Tons de Cinza", img_gray)
cv2.imshow("Passa-Baixa: Media 9x9", img_media)
cv2.imshow("Passa-Baixa: Gaussiano 9x9", img_gauss)
cv2.imshow("Passa-Alta: Sobel Magnitude", sobel_total_abs)
cv2.imshow("Passa-Alta: Laplaciano", laplaciano_abs)

print("Pressione qualquer tecla em uma das janelas para encerrar...")
cv2.waitKey(0)
cv2.destroyAllWindows()
```
