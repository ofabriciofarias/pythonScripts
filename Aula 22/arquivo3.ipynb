```python
import cv2
import numpy as np

# 1. Carrega e suaviza com Passa-Baixa (elimina textura de pelos)
img = cv2.imread("bentoespock.jpeg")
cinza = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
suave = cv2.GaussianBlur(cinza, (7, 7), 0)

# 2. Passa-Alta: Detecção de Bordas (Canny, que combina Sobel + Supressão de não-máximos)
bordas = cv2.Canny(suave, threshold1=30, threshold2=100)

# 3. Morfologia para fechar pequenos vãos nas bordas encontradas
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
bordas_fechadas = cv2.morphologyEx(bordas, cv2.MORPH_CLOSE, kernel)

# 4. Transforma o mapa de píxeis em dados vetoriais (Contornos)
contornos, _ = cv2.findContours(bordas_fechadas, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

img_resultado = img.copy()

# 5. O que fazemos com os contornos? Filtramos por área e extraímos Bounding Boxes
for c in contornos:
    area = cv2.contourArea(c)
    # Filtra ruídos pequenos e pega apenas componentes expressivos
    if area > 3000:
        x, y, w, h = cv2.boundingRect(c)
        # Desenha a caixa delimitadora encontrada pelo algoritmo clássico
        cv2.rectangle(img_resultado, (x, y), (x + w, y + h), (0, 255, 0), 2)
        cv2.putText(img_resultado, f"Area: {int(area)}", (x, y - 5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)

cv2.imwrite("deteccao_via_filtros.jpg", img_resultado)
```
