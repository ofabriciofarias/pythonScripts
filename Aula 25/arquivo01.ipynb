```python
# Resolve o conflito de bibliotecas (KMP) que derruba o kernel no Windows
#import os
#os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"

# ==============================================================================
# SLIDE 1: IMPORTAÇÃO E VERIFICAÇÃO DA RTX 4070
# ==============================================================================
from ultralytics import YOLO
import torch

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Hardware ativado com sucesso: {device.upper()} -> {torch.cuda.get_device_name(0)}")

# ==============================================================================
# SLIDE 2: CARREGAMENTO DO MODELO PRÉ-TREINADO
# ==============================================================================
modelo = YOLO("yolov8n.pt")

# ==============================================================================
# SLIDE 3: TREINAMENTO E EXTRAÇÃO DE MÉTRICAS (PRECISÃO, RECALL, F1/mAP)
# ==============================================================================
if __name__ == '__main__':
    # Executa o treino
    resultados = modelo.train(
        #data="libras.v21i.yolov8/data.yaml",
        data = "libras.v21i.yolov8/data.yaml",
        epochs=100,           
        imgsz=640,           # Resolução padrão alta suportada pela RTX 4070
        batch=16,            # Lote ideal para a VRAM da sua placa
        device=device,       # Utiliza a RTX 4070
        workers=0,           # Mantido em 0 para garantir estabilidade no Jupyter/Windows
        name="libras_yolov8"
    )
    
    print("\nTreinamento concluído com sucesso na RTX 4070!")
    
    # ==========================================================================
    # SLIDE 4: AVALIAÇÃO E EXIBIÇÃO DAS MÉTRICAS FINAIS
    # ==========================================================================
    print("\nA carregar os melhores pesos ('best.pt') para validação...")
    
    # Carrega explicitamente o melhor modelo gerado durante o treino
    modelo_treinado = YOLO("runs/detect/libras_yolov8/weights/best.pt")
    
    # Executa a validação no conjunto de teste/validação do dataset
    metricas = modelo_treinado.val()
    
    # Extração das métricas principais fornecidas pelo ecossistema YOLOv8
    precisao_media = metricas.box.p    # Precisão (Precision) por classe ou média
    recall_medio = metricas.box.r      # Revocação (Recall) por classe ou média
    map50 = metricas.box.map50         # mAP com limiar de IoU de 0.50 (equivalente à acurácia global de detecção)
    map50_95 = metricas.box.map        # mAP com limiar de IoU de 0.50 a 0.95
    
    print("\n" + "="*50)
    print("MÉTRICAS DE DESEMPENHO DO MODELO LIBRAS:")
    print("="*50)
    print(f"• Precisão Média (Precision): {precisao_media.mean():.4f}")
    print(f"• Revocação Média (Recall):   {recall_medio.mean():.4f}")
    print(f"• mAP@0.5 (Acurácia de Caixa): {map50:.4f}")
    print(f"• mAP@0.5:0.95 (Geral):        {map50_95:.4f}")
    print("="*50)
```
