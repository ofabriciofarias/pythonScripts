```python
# ==============================================================================
# RECONHECIMENTO DE LIBRAS (ALFABETO MANUAL) COM MEDIAPIPE
# ==============================================================================
import cv2
import mediapipe as mp
import math

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

hands = mp_hands.Hands(
    static_image_mode=False,
    max_num_hands=1,
    min_detection_confidence=0.7,
    min_tracking_confidence=0.7
)

# Função auxiliar para calcular distância euclidiana entre dois pontos do esqueleto
def calcular_distancia(p1, p2):
    return math.sqrt((p1.x - p2.x)**2 + (p1.y - p2.y)**2)

cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)

print("Iniciando reconhecimento de Libras. Pressione 'q' para sair.")

while cap.isOpened():
    sucesso, frame = cap.read()
    if not sucesso:
        break

    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    resultado = hands.process(frame_rgb)

    letra_reconhecida = "Aguardando sinal..."

    if resultado.multi_hand_landmarks:
        for hand_landmarks in resultado.multi_hand_landmarks:
            mp_drawing.draw_landmarks(frame, hand_landmarks, mp_hands.HAND_CONNECTIONS)

            # Coordenadas de pontos-chave
            # Pontas dos dedos: Polegar(4), Indicador(8), Médio(12), Anelar(16), Mínimo(20)
            # Articulações base: Polegar(2), Indicador(5), Médio(9), Anelar(13), Mínimo(17)
            lm = hand_landmarks.landmark

            # Exemplo prático de regra geométrica simples:
            # Verificar se o indicador está levantado e os outros abaixados (ex: letra 'I' ou similar)
            dedo_indicador_levantado = lm[8].y < lm[6].y
            dedo_medio_baixado = lm[12].y > lm[10].y
            dedo_anelar_baixado = lm[16].y > lm[14].y
            dedo_minimo_baixado = lm[20].y > lm[18].y

            if dedo_indicador_levantado and dedo_medio_baixado and dedo_anelar_baixado and dedo_minimo_baixado:
                letra_reconhecida = "Letra provavel: L / 1"
            else:
                letra_reconhecida = "Mao detetada (Posicao neutra)"

    # Escreve o resultado detetado na tela
    cv2.putText(frame, f"Sinal: {letra_reconhecida}", (30, 50), 
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2, cv2.LINE_AA)

    cv2.imshow("Reconhecedor de Libras - MediaPipe", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
